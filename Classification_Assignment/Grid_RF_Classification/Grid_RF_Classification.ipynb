```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("CKD.csv")
```

```python
data.head()
```

```python
data['classification'].value_counts()
```

```python
data.isnull().sum()
```

```python
data=pd.get_dummies(data,drop_first=True)
```

```python
data=data.astype(int)
```

```python
indep=data.iloc[:,0:27].values
```

```python
dep=data['classification_yes']
```

```python
indep
```

```python
dep
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(indep,dep,test_size=0.3,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {'criterion':['gini','entropy'],
              'max_features': ['auto','sqrt','log2'],
              'n_estimators':[10,100]} 

grid = GridSearchCV(RandomForestClassifier(), param_grid, refit = True, verbose = 3,n_jobs=-1,scoring='f1') 
   
grid.fit(x_train, y_train) 
 
```

```python
# print best parameter after tuning 
#print(grid.best_params_) 
re=grid.cv_results_
#print(re)
grid_predictions = grid.predict(x_test) 
   
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, grid_predictions)


# print classification report 
from sklearn.metrics import classification_report
clf_report = classification_report(y_test, grid_predictions)
```

```python
print(grid.best_params_)
```

```python
table=pd.DataFrame.from_dict(re)
```

```python
table
```

```python
from sklearn.metrics import f1_score
f1_macro=f1_score(y_test,grid_predictions,average='weighted')
print("The f1_macro value for best parameter {}:".format(grid.best_params_),f1_macro)
```
