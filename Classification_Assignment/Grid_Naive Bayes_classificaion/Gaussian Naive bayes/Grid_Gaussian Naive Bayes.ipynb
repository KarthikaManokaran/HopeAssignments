```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
data= pd.read_csv("CKD.csv")
```

```python
data
```

```python
data['classification'].value_counts()
```

```python
data=pd.get_dummies(data,drop_first=True)
data=data.astype(int)
data
```

```python
indep=data.iloc[:,0:27].values
dep=data['classification_yes']
```

```python
indep
```

```python
dep
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test= train_test_split(indep,dep,test_size=0.3,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
#Gaussian Naive bayes - got error ( ValueError: Negative values in data passed to MultinomialNB (input X))
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.naive_bayes import GaussianNB

# Create an untuned KNN model
clf = GaussianNB()

# Define the parameter grid
param_grid = {
    'var_smoothing': np.logspace(0, -9, num=100)
}

# Create a GridSearchCV object
grid = GridSearchCV(clf, param_grid, cv=5)

# Fit the grid search object to the data
grid.fit(x_train, y_train)
```

```python
best_params = grid.best_params_
```

```python
print(best_params)
```

```python
re=grid.cv_results_
```

```python
grid_predictions=grid.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,grid_predictions)
```

```python
from sklearn.metrics import classification_report
clf_report = classification_report(y_test, grid_predictions)
```

```python
print(clf_report)
```

```python
print(cm)
```

```python
from sklearn.metrics import f1_score
f1_macro=f1_score(y_test,grid_predictions,average='weighted')
print("The f1_macro value for best parameter {}:".format(grid.best_params_),f1_macro)
```

```python
from sklearn.metrics import roc_auc_score

roc_auc_score(y_test,grid.predict_proba(x_test)[:,1])
```
