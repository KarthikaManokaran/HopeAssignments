# independent sample - unpaired T-Test

Different Group(Male,Female) but same conditions (salary)

```python
from scipy.stats import ttest_ind
dataset=dataset.dropna()
male=dataset[dataset["gender"]=="M"]["salary"]
female=dataset[dataset["gender"]=="F"]["salary"]
print (male)
```

```python
ttest_ind(male,female)
```

# Dependent sample -paired T-Test

Same Group(Male) but different conditions (ssc_p and hsc_p)

```python
from scipy.stats import ttest_rel
male1=dataset[dataset["gender"]=="M"]["ssc_p"]
male2=dataset[dataset["gender"]=="M"]["hsc_p"]
ttest_rel(male1,male2)

```

# Hypothesis Testing

Statement : Test whether ssc_p of male and hsc_p of male is same? 

Null hypothesis : There is no similarity between 10th pass mark of male and 12th pass mark of male.

```python
from scipy.stats import ttest_rel
male1=dataset[dataset["gender"]=="M"]["ssc_p"]
male2=dataset[dataset["gender"]=="M"]["hsc_p"]
ttest_rel(male1,male2)
```

```python
# Reject null hypothesis when the p value is <0.05 .Here p value is > 0.05 .so that, reject alternate hypothesis and accept null hypothesis.
```
